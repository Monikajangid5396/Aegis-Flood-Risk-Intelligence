```python
X_hist_train,X_hist_test,y_hist_train,y_hist_test = train_test_split(
    X_historical,
    y_historical,
    test_size=0.20,
    random_state=42
)
print("Training data:")
print("X:",X_hist_train.shape)
print("y:",y_hist_train.shape)
print("\nTesting data:")
print("X:",X_hist_test.shape)
print("y:",y_hist_test.shape)
```

```python
historical_scaler = StandardScaler()
X_hist_train_scaled = historical_scaler.fit_transform(
    X_hist_train
)
X_hist_test_scaled = historical_scaler.transform(
    X_hist_test
)
print("Scaled training shape:",X_hist_train_scaled.shape)
print("Scaled testing shape:",X_hist_test_scaled.shape)
```

```python
historical_ridge = Ridge(alpha=1.0)

historical_ridge.fit(
    X_hist_train_scaled,
    y_hist_train
)

hist_ridge_train_pred = historical_ridge.predict(
    X_hist_train_scaled
)

hist_ridge_test_pred = historical_ridge.predict(
    X_hist_test_scaled
)

hist_ridge_train_r2 = r2_score(
    y_hist_train,
    hist_ridge_train_pred
)

hist_ridge_test_r2 = r2_score(
    y_hist_test,
    hist_ridge_test_pred
)

hist_ridge_train_mae = mean_absolute_error(
    y_hist_train,
    hist_ridge_train_pred
)

hist_ridge_test_mae = mean_absolute_error(
    y_hist_test,
    hist_ridge_test_pred
)

hist_ridge_train_rmse = np.sqrt(
    mean_squared_error(
        y_hist_train,
        hist_ridge_train_pred
    )
)

hist_ridge_test_rmse = np.sqrt(
    mean_squared_error(
        y_hist_test,
        hist_ridge_test_pred
    )
)

print("Historical Ridge Regression")
print("---------------------------")

print("Training R²:",
      round(hist_ridge_train_r2, 4))

print("Testing R² :",
      round(hist_ridge_test_r2, 4))

print("\nTraining MAE:",
      round(hist_ridge_train_mae, 4))

print("Testing MAE :",
      round(hist_ridge_test_mae, 4))

print("\nTraining RMSE:",
      round(hist_ridge_train_rmse, 4))

print("Testing RMSE :",
      round(hist_ridge_test_rmse, 4))

print(
    "\nOverfitting gap:",
    round(
        hist_ridge_train_r2 - hist_ridge_test_r2,
        4
    )
)
```

```python
risk_data = historical_flood_weather.copy()

q1 = risk_data["flood_affected_percent"].quantile(0.25)
median = risk_data["flood_affected_percent"].quantile(0.50)
q3 = risk_data["flood_affected_percent"].quantile(0.75)

risk_data["risk_category"] = pd.cut(
    risk_data["flood_affected_percent"],
    bins=[
        -np.inf,
        q1,
        median,
        q3,
        np.inf
    ],
    labels=[
        "Low",
        "Moderate",
        "High",
        "Very High"
    ],
    include_lowest=True
)

print("Risk thresholds:")
print("Low       :", f"≤ {q1:.4f}%")
print("Moderate  :", f"> {q1:.4f}% to ≤ {median:.4f}%")
print("High      :", f"> {median:.4f}% to ≤ {q3:.4f}%")
print("Very High :", f"> {q3:.4f}%")

print("\nRisk category counts:")
print(
    risk_data["risk_category"]
    .value_counts()
    .reindex(["Low", "Moderate", "High", "Very High"])
)

print("\nFirst 10 records:")
display(
    risk_data[
        [
            "district",
            "sub_district",
            "flood_affected_percent",
            "risk_category"
        ]
    ].head(10)
)
```
